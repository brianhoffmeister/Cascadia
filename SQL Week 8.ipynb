```python
import sqlite3 #imports sqlite3 tool
import pandas as pd #imports pandas with the pd nickname
conn = sqlite3.connect('my_data.db') #creates a database and connects sqlite3 to it
cursor = conn.cursor() #creates a cursor 
```

**Task 1 - Sales**

```python
#Display sale_date, product, and price from the sales table.
pd.read_csv('sales.csv').to_sql('sales', conn, if_exists = 'replace', index = False) #reads in the sales csv file and connects to sql

pd.read_sql_query('''
SELECT sale_date, product, price FROM sales
''', conn)
```

**Task 2 - HR**

```python
#Find employees in Chicago with more than 10 years of experience using WHERE and AND.
pd.read_csv('hr_employees.csv').to_sql('employees', conn, if_exists = 'replace', index = False) #reads in the hr_employee csv file and connects to sql

pd.read_sql_query('''
SELECT name, city, department, salary
FROM employees
WHERE city = 'Chicago' and years_experience > 10
''', conn) #shows employees from Chicago with more then 10 years of experience
```

**Task 3 - Real Estate**

```python
#Display the three largest properties by square footage using ORDER BY and LIMIT.
pd.read_csv('realestate_listings.csv').to_sql('listings', conn, if_exists = 'replace', index = False) #reads in the real estate csv file and connects to sql

pd.read_sql_query('''
SELECT listing_id, neighborhood, sqft, list_price
FROM listings
ORDER BY sqft DESC
LIMIT 3
''', conn) #selects the needed columns from the query and shows top 3 ordering them by sqft in descending order 
```

**Task 4 - Marketing**

```python
#Calculate total weekly impressions using SUM() and GROUP BY. Adding in total spend by week as well to give an impression of spend ROI
pd.read_csv('marketing_campaigns.csv').to_sql('campaigns', conn, if_exists = 'replace', index = False)

pd.read_sql_query('''
SELECT week, SUM(impressions) AS total_impressions, SUM(spend) AS total_spend
FROM campaigns
GROUP BY week
''', conn)
```

**Task 5 - Restaurant**

```python
#Find dates when total restaurant sales exceeded $400.
pd.read_csv('restaurant_orders.csv').to_sql('orders', conn, if_exists = 'replace', index = False)

pd.read_sql_query('''
SELECT order_date, SUM(order_total) AS total_revenue
FROM orders
GROUP BY order_date
HAVING SUM(order_total) > 400
''', conn)
```

**Task 6 - Healthcare**

```python
#Find the average patient age and longest waiting time for each doctor.
pd.read_csv('clinic_visits.csv').to_sql('visits', conn, if_exists = 'replace', index = False)

pd.read_sql_query('''
SELECT COUNT(*) AS total_visits,
ROUND(AVG(patient_age), 1) AS avg_patient_age,
MAX(wait_minutes) AS longest_wait
FROM visits
''', conn)
```

**Task 7 - Banking**

```python
#Join customer and transaction tables to calculate total transaction amounts by branch.
pd.read_csv('bank_customers.csv').to_sql('customers', conn, if_exists = 'replace', index = False)
pd.read_csv('bank_transactions.csv').to_sql('transactions', conn, if_exists = 'replace', index = False)

pd.read_sql_query('''
SELECT c.branch, SUM(t.amount) AS total_transactions
FROM transactions AS t
JOIN customers AS c ON t.customer_id = c.customer_id
GROUP BY c.branch
''', conn)
```
